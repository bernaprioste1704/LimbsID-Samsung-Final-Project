## Preparation

```python
import pandas as pd
```

```python
df = pd.read_csv('mpii_human_pose.csv')
df.info()
```

```python
df.head()
```

## Distribution of The Scale of images

```python
df["Scale"].plot(kind='hist', bins=100)
```

```python
# Create a dictionary to store the counts
counts = {
	"r_ankle": (df["r ankle_X"] > 0).sum(),
	"l_ankle": (df["l ankle_X"] > 0).sum(),
	"r_knee": (df["r knee_X"] > 0).sum(),
	"l_knee": (df["l knee_X"] > 0).sum(),
	"r_hip": (df["r hip_X"] > 0).sum(),
	"l_hip": (df["l hip_X"] > 0).sum(),
	"pelvis": (df["pelvis_X"] > 0).sum(),
	"thorax": (df["thorax_X"] > 0).sum(),
	"upper_neck": (df["upper neck_X"] > 0).sum(),
	"head_top": (df["head top_X"] > 0).sum(),
	"r_wrist": (df["r wrist_X"] > 0).sum(),
	"l_wrist": (df["l wrist_X"] > 0).sum(),
	"r_elbow": (df["r elbow_X"] > 0).sum(),
	"l_elbow": (df["l elbow_X"] > 0).sum(),
	"r_shoulder": (df["r shoulder_X"] > 0).sum(),
	"l_shoulder": (df["l shoulder_X"] > 0).sum()
}

# Convert the dictionary to a dataframe
df2 = pd.DataFrame(list(counts.items()), columns=["Joint", "Count"])
df2

```

## Distribution of the number of keypoints

```python
df2.plot(kind='bar', x='Joint', y='Count')
```

There is no keypoint that is in all of the photos
